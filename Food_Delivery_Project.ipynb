```python
import pandas as pd

# Load the file
df = pd.read_csv("Food_Delivery_Route_Efficiency_Dataset.csv")

# Display frist 5 orders
print("FOOD DELIVERY ORDERS RECORDS")
print(df.head())

# Total Orders-200 rows
print("\nTotal Orders:", len(df))

# Average Delivery Time
print("Average Delivery Time:", df["delivery_time_min"].mean())

# Average Delivery Time by Traffic Level
print("\nTraffic-wise Average Delivery Time:")
print(df.groupby("traffic_level")["delivery_time_min"].mean())

#  Average Delivery Time Weather
print("\nWeather-wise Average Delivery Time:")
print(df.groupby("weather")["delivery_time_min"].mean())

# Orders by Delivery Mode
print("\nDelivery Mode-wise Orders:")
print(df.groupby("delivery_mode")["order_id"].count())

# view data for one Zone
zone = input("Enter customer zone name: ")
print(df[df["customer_zone"] == zone])
```
